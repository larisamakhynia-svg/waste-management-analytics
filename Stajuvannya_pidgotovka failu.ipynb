```python
import pandas as pd

# 1. Завантаження файлу
raw_file = "waste_data.csv"
df = pd.read_csv(raw_file)

# 2. Відбір суттєвих колонок (без колонки 'Спеціальне значення даних')
cols_to_keep = [
    'Вихідний статистичний показник',
    'Код місцезнаходження за КАТОТТГ',
    'Вид установок для оброблення відходів',
    'Тип відходів, що видаляється',
    'Стан використання полігонів',
    'Період часу',
    'Одиниця вимірювання',
    'Значення спостереження',
    'Спеціальне значення даних'  # залишаємо тимчасово для обробки 'Явищ не було'
]
df_clean = df[cols_to_keep].copy()

# 3. Перейменування колонок
df_clean.columns = [
    'Показник', 'Регіон', 'Вид_установок', 'Тип_відходів',
    'Стан_використання', 'Рік', 'Одиниця_виміру', 'Значення', 'Спеціальне_значення'
]

# 4. Обрізка пробілів, заміна 'Явищ не було' на 0 та видалення колонки 'Спеціальне_значення'
for col in df_clean.select_dtypes(include='object').columns:
    df_clean[col] = df_clean[col].astype(str).str.strip().replace({'nan': None})

df_clean.loc[df_clean['Спеціальне_значення'] == 'Явищ не було', 'Значення'] = 0.0

# Остаточне видалення колонки 'Спеціальне_значення'
df_clean = df_clean.drop(columns=['Спеціальне_значення'])

# 5. Збереження очищеного CSV
df_clean.to_csv("cleaned_waste_facilities_2025_final.csv", index=False, encoding='utf-8-sig')
print("Файл успішно очищено та збережено!")
```

```python
import pandas as pd
import numpy as np

# Зчитуємо файл
df = pd.read_csv('cleaned_waste_facilities_2025_final.csv')

# Перевіряємо перші 5 рядків та унікальні значення у Показниках
print(df.head())
print("\nУнікальні показники:")
print(df['Показник'].unique())
```

```python
# Виключаємо сумарний рядок "Україна", щоб не спотворювати регіональну статистику
df_regions = df[df['Регіон'] != 'Україна'].copy()

# Розрахунок метрик за кожним Показником
def get_stats(group):
    p25 = group['Значення'].quantile(0.25)
    p75 = group['Значення'].quantile(0.75)
    mode_val = group['Значення'].mode()
    
    return pd.Series({
        'COUNT': group['Значення'].count(),
        'SUM': group['Значення'].sum(),
        'AVG': group['Значення'].mean(),
        'MEDIAN': group['Значення'].median(),
        'MODE': mode_val.iloc[0] if not mode_val.empty else np.nan,
        'MIN': group['Значення'].min(),
        'MAX': group['Значення'].max(),
        'STDDEV': group['Значення'].std(),
        'p25': p25,
        'p75': p75,
        'IQR': p75 - p25
    })

stats_summary = df_regions.groupby('Показник').apply(get_stats, include_groups=False)
stats_summary
```

```python
# Перетворення з Long у Wide формат за Регіонами та Показниками
pivot_df = df_regions.pivot_table(
    index='Регіон', 
    columns=['Показник', 'Стан_використання'], 
    values='Значення', 
    aggfunc='sum'
).fillna(0)

# Виведення результату
pivot_df.head()
```

```python
# 1. Спрощуємо назви колонок (прибираємо MultiIndex)
pivot_df.columns = [f"{col[0]}_{col[1]}" for col in pivot_df.columns]
pivot_df = pivot_df.reset_index()

# 2. Розраховуємо похідні метрики для аналітики
# Заповненість об'єму (%)
pivot_df["Заповненість_обєму_%"] = (
    1
    - (
        pivot_df["Об'єм полігонів для видалення відходів_Залишок"]
        / pivot_df["Об'єм полігонів для видалення відходів_По проекту"]
    )
) * 100

# Заповненість площі (%)
pivot_df["Заповненість_площі_%"] = (
    1
    - (
        pivot_df["Площа полігонів для видалення відходів_Залишок"]
        / pivot_df["Площа полігонів для видалення відходів_По проекту"]
    )
) * 100

# Середній об'єм полігону в регіоні
pivot_df["Середній_обєм_полігону"] = (
    pivot_df["Об'єм полігонів для видалення відходів_По проекту"]
    / pivot_df["Кількість полігонів для видалення відходів_Не застосовується"]
)

# Середня потужність 1 установки
pivot_df["Середня_потужність_1_установки"] = (
    pivot_df[
        "Потужність установок для оброблення відходів_Не застосовується"
    ]
    / pivot_df["Кількість установок для оброблення відходів_Не застосовується"]
).fillna(0)

# 3. Експорт файлів для Tableau
pivot_df.to_csv("tableau_regional_metrics.csv", index=False, encoding="utf-8-sig")
stats_summary.to_csv("tableau_descriptive_stats.csv", encoding="utf-8-sig")

print("Файли успішно збережені!")
```
